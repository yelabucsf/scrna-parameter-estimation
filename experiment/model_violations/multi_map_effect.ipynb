# What does multi-mapped reads do to the variance estimator?

```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
```

### Typical case

```python
def variance_estimator(Y):

    return np.mean((Y**2-Y*(1-q)))/q**2 - (np.mean(Y)**2/q**2)

# Normal framework
q = 0.05
N = 1000000
X1 = stats.poisson.rvs(50, size=N)
X2 = stats.poisson.rvs(30, size=N)

Y1 = stats.binom.rvs(X1, p=q)
Y2 = stats.binom.rvs(X2, p=q)
```

```python
variance_estimator(Y1)
```

```python
variance_estimator(Y2)
```

### Ambiguity but no multimapping

```python
q = 0.05
N = 1000000
X1 = stats.poisson.rvs(50, size=N)
X2 = stats.poisson.rvs(80, size=N)

ambiguity = 0.005 # probability each UMI is assigned to another 
Y1 = stats.binom.rvs(X1, p=q)
Y2 = stats.binom.rvs(X2, p=q)

Y1 = (1-ambiguity)*Y1 + ambiguity*Y2
Y2 = (1-ambiguity)*Y2 + ambiguity*Y1

# Y1 = stats.binom.rvs(X1, p=q)
# Y2 = stats.binom.rvs(X2, p=q)
```

```python
variance_estimator(Y1)
```
